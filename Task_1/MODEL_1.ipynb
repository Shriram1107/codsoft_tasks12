```python
!pip install -q pandas scikit-learn
import pandas as pd
import re

from google.colab import files

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report

print("Upload the following files:")
print("1. train_data.txt")
print("2. test_data.txt")
print("3. test_data_solution.txt")

uploaded = files.upload()

train_file = "train_data.txt"
test_file = "test_data.txt"
solution_file = "test_data_solution.txt"

train = pd.read_csv(
    train_file,
    sep=" ::: ",
    engine="python",
    names=["ID", "TITLE", "GENRE", "DESCRIPTION"]
)

test = pd.read_csv(
    test_file,
    sep=" ::: ",
    engine="python",
    names=["ID", "TITLE", "DESCRIPTION"]
)

solution = pd.read_csv(
    solution_file,
    sep=" ::: ",
    engine="python",
    names=["ID", "TITLE", "GENRE", "DESCRIPTION"]
)

# Remove the first row if it contains column names
if train.iloc[0]["ID"] == "ID":
    train = train.iloc[1:].reset_index(drop=True)

if test.iloc[0]["ID"] == "ID":
    test = test.iloc[1:].reset_index(drop=True)

if solution.iloc[0]["ID"] == "ID":
    solution = solution.iloc[1:].reset_index(drop=True)

# ===========================
# STEP 5: Dataset Information
# ===========================

print("\n==============================")
print("TRAINING DATASET")
print("==============================")

print(train.head())

print("\nTraining Shape :", train.shape)

print("\nMissing Values")
print(train.isnull().sum())

# ===========================
# STEP 6: Data Cleaning
# ===========================

def clean_text(text):

    text = str(text).lower()

    text = re.sub(r'[^a-zA-Z ]', ' ', text)

    text = re.sub(r'\s+', ' ', text)

    return text.strip()

train["DESCRIPTION"] = train["DESCRIPTION"].apply(clean_text)
test["DESCRIPTION"] = test["DESCRIPTION"].apply(clean_text)
solution["DESCRIPTION"] = solution["DESCRIPTION"].apply(clean_text)

# ===========================
# STEP 7: Build Model
# ===========================

model = Pipeline([

    ("tfidf",
     TfidfVectorizer(
         stop_words="english",
         max_features=50000
     )
    ),

    ("classifier",
     LogisticRegression(
         max_iter=1000
     )
    )

])

# ===========================
# STEP 8: Train Model
# ===========================

print("\nTraining the model...")

model.fit(train["DESCRIPTION"], train["GENRE"])

print("Model training completed.")

# ===========================
# STEP 9: Predict Genres
# ===========================

predictions = model.predict(test["DESCRIPTION"])

# ===========================
# STEP 10: Evaluate Model
# ===========================

accuracy = accuracy_score(solution["GENRE"], predictions)

print("\n======================================")
print("MODEL ACCURACY :", round(accuracy * 100,2), "%")
print("======================================")

print("\nClassification Report\n")

print(classification_report(solution["GENRE"], predictions))

# ===========================
# STEP 11: Create Prediction File
# ===========================

submission = pd.DataFrame({

    "Movie_ID": test["ID"],

    "Movie_Title": test["TITLE"],

    "Predicted_Genre": predictions

})

print("\nSample Predictions\n")

print(submission.head(10))

submission.to_csv("movie_genre_predictions.csv", index=False)

print("\nPrediction file created successfully!")

# ===========================
# STEP 12: Download Prediction File
# ===========================

files.download("movie_genre_predictions.csv")

print("\nProject Completed Successfully!")
```
